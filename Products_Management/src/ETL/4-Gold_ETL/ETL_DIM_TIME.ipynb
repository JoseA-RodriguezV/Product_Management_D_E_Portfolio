# **ETL GOLD DIM TIME**

```python
%sql
merge into products.gold.dim_time as target
using (
     select distinct
    cast(DATE_FORMAT(start_date_market, 'yyyyMMdd') AS BIGINT) as date_id,
    start_date_market,
    end_date_market,
    year(start_date_market) as market_year,
    month(start_date_market) as market_month,
    QUARTER(start_date_market) as market_quarter,
    case
      when market_month <= 6 then 1
      else 2
    end as market_semester,
    weekofyear(start_date_market) as market_week
  from
    products.silver.product_list_silver
) as source
on target.date_id = source.date_id
--when matched then update set *
when not matched then insert (date_id,
  start_market_date,
  end_market_date,
  market_year,
  market_month,
  market_quarter,
  market_semester,
  week_number
)
values(
    source.date_id,
    source.start_date_market,
    source.end_date_market,
    source.market_year,
    source.market_month,
    source.market_quarter,
    source.market_semester,
    source.market_week
)
```
