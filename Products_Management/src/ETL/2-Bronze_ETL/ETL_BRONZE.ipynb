```python
%sql
merge into
  products.bronze.product_list_markets_bronze as target
using (
  select
    nombre_producto,
    unidad_medida,
    precio_inicio,
    precio_fin,
    nombre_mercado,
    inicio_fecha_precios,
    fin_fecha_precios,
    error
  from
    products.landing.product_list_markets_landing
) as source
on
  target.nombre_producto = source.nombre_producto
  and target.unidad_medida = source.unidad_medida
  and target.precio_inicio = source.precio_inicio
  and target.precio_fin = source.precio_fin
  and target.nombre_mercado = source.nombre_mercado
  and target.inicio_fecha_precios = source.inicio_fecha_precios
  and target.fin_fecha_precios = source.fin_fecha_precios
when not matched then insert (
    nombre_producto,
    unidad_medida,
    precio_inicio,
    precio_fin,
    nombre_mercado,
    inicio_fecha_precios,
    fin_fecha_precios,
    error
  )
  values (
    source.nombre_producto,
    source.unidad_medida,
    source.precio_inicio,
    source.precio_fin,
    source.nombre_mercado,
    source.inicio_fecha_precios,
    source.fin_fecha_precios,
    source.error
  )
```
