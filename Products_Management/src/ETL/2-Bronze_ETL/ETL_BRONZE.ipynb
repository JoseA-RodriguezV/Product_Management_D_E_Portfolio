```python
%sql
merge into
  products.bronze.product_list_markets_bronze as target
using (
  select
    trim(nombre_producto) as nombre_producto,
    trim(unidad_medida) as unidad_medida,
    trim(precio_inicio) as precio_inicio,
    trim(precio_fin) as precio_fin,
    trim(nombre_mercado) as nombre_mercado,
    trim(inicio_fecha_precios) as inicio_fecha_precios,
    trim(fin_fecha_precios) as fin_fecha_precios,
    trim(error) as error
  from
    products.landing.product_list_markets_landing
  where
    nombre_producto is not null 
) as source
on
  coalesce(target.nombre_producto, '') = coalesce(source.nombre_producto, '')
  and coalesce(target.unidad_medida, '') = coalesce(source.unidad_medida, '')
  and coalesce(target.precio_inicio, '') = coalesce(source.precio_inicio, '')
  and coalesce(target.precio_fin, '') = coalesce(source.precio_fin, '')
  and coalesce(target.nombre_mercado, '') = coalesce(source.nombre_mercado, '')
  and coalesce(target.inicio_fecha_precios, '') = coalesce(source.inicio_fecha_precios, '')
  and coalesce(target.fin_fecha_precios, '') = coalesce(source.fin_fecha_precios, '')
when not matched then insert (
    nombre_producto,
    unidad_medida,
    precio_inicio,
    precio_fin,
    nombre_mercado,
    inicio_fecha_precios,
    fin_fecha_precios,
    error
  )
  values (
    source.nombre_producto,
    source.unidad_medida,
    source.precio_inicio,
    source.precio_fin,
    source.nombre_mercado,
    source.inicio_fecha_precios,
    source.fin_fecha_precios,
    source.error
  );
```
