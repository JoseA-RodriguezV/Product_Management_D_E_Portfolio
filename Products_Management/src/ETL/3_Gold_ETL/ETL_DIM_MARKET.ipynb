# **ETL GOLD DIM MARKET**

```python
%sql
merge into
  products.gold.dim_market as target
using (
  select distinct
    market_name,
    city,
    department
  from
    products.silver.product_list_silver
) as source
on
  target.market_name = source.market_name
  and target.city = source.city
  and target.department = source.department
when matched then update set
  target.market_name = source.market_name,
  target.city = source.city,
  target.department = source.department
when not matched then insert (market_name, city, department) values (market_name, city, department)
```
