# **ETL GOLD DIM PRODUCT**

```python
%sql
 select distinct
    product_name,
    case
      when
        product_name like '%res%'
        or product_name like '%pollo%'
        or product_name like '%cerdo%'
      then
        'carnes'
      when product_name like '%pescado%' then 'pescados y mariscos'
      when
        product_name like '%leche%'
        or product_name like '%ques%'
        or product_name like '%yog%'
        or product_name like '%mantequill%'
        or product_name like '%hue%'
      then
        'lacteos y huevos'
      when
        product_name like '%arroz%'
        or product_name like '%frijo%'
        or product_name like '%maíz%'
      then
        'granos básicos'
      when
        product_name like '%ceboll%'
        or product_name like '%tomate%'
        or product_name like '%papa%'
        or product_name like '%yuca%'
        or product_name like '%repollo%'
        or product_name like '%zanahoria%'
      then
        'verduras'
      when
        product_name like '%naranja%'
        or product_name like '%plátano%'
        or product_name like '%manzana%'
        or product_name like '%banano%'
      then
        'frutas'
      when
        product_name like '%espaguetti%'
        or product_name like '%azúcar%'
        or product_name like '%café%'
        or product_name like '%salsa%'
        or product_name like '%aceite%'
        or product_name like '%manteca%'
        or product_name like '%sal%'
        or product_name like '%tortilla%'
        or product_name like '%pan%'
      then
        'despensa'
      when product_name like '%jugo%' then 'bebida'
      else 'otros'
    end as product_category,
    product_unit
  from
    products.silver.product_list_silver
```

```python
%sql
merge into
  products.gold.dim_product as target
using (
  select distinct
    product_name,
    case
      when
        product_name like '%res%'
        or product_name like '%pollo%'
        or product_name like '%cerdo%'
      then
        'carnes'
      when product_name like '%pescado%' then 'pescados y mariscos'
      when
        product_name like '%leche%'
        or product_name like '%ques%'
        or product_name like '%yog%'
        or product_name like '%mantequill%'
        or product_name like '%hue%'
      then
        'lacteos y huevos'
      when
        product_name like '%arroz%'
        or product_name like '%frijo%'
        or product_name like '%maíz%'
      then
        'granos básicos'
      when
        product_name like '%ceboll%'
        or product_name like '%tomate%'
        or product_name like '%papa%'
        or product_name like '%yuca%'
        or product_name like '%repollo%'
        or product_name like '%zanahoria%'
      then
        'verduras'
      when
        product_name like '%naranja%'
        or product_name like '%plátano%'
        or product_name like '%manzana%'
        or product_name like '%banano%'
      then
        'frutas'
      when
        product_name like '%espaguetti%'
        or product_name like '%azúcar%'
        or product_name like '%café%'
        or product_name like '%salsa%'
        or product_name like '%aceite%'
        or product_name like '%manteca%'
        or product_name like '%sal%'
        or product_name like '%tortilla%'
        or product_name like '%pan%'
      then
        'despensa'
      when product_name like '%jugo%' then 'bebida'
      else 'otros'
    end as product_category,
    product_unit
  from
    products.silver.product_list_silver
) as source
on
  target.product_name = source.product_name
  and target.product_category = source.product_category
  and target.product_unit = source.product_unit
when matched then update set
  target.product_name = source.product_name,
  target.product_category = source.product_category,
  target.product_unit = source.product_unit
when not matched then insert (product_name, product_category, product_unit)
  values (source.product_name, source.product_category, source.product_unit)
```
