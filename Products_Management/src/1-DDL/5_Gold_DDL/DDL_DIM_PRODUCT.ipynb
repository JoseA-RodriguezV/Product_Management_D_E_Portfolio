# **PRODUCT DIMENSION**

DDL for Product Dimension Table in Gold Layer

```python
%sql
DROP TABLE IF EXISTS products.gold.dim_product;

create table products.gold.dim_product (
    product_id BIGINT
      GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1)
      PRIMARY KEY
      COMMENT 'Product ID',
    product_name STRING NOT NULL COMMENT 'Product name description',
    product_category STRING NOT NULL COMMENT 'Product category type',
    product_unit STRING COMMENT 'Product Unit of measurement value',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
  ) USING DELTA
  TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
  COMMENT 'Product Dimension - Star Schema';
```
