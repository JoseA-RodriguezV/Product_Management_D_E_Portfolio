# **Creating Silver Table**

DDL for Silver Layer

```python
%sql
drop table if EXISTS products.silver.product_list_silver;

create table products.silver.product_list_silver (
    product_id BIGINT
      GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1)
      PRIMARY KEY
      COMMENT 'Auto incremental Primary Key for Products.',
    product_name STRING COMMENT 'Product Name description.',
    product_unit STRING COMMENT 'Product Unit of measurement value.',
    start_date_price DECIMAL(10, 2) COMMENT 'Product price on the start date on the market.',
    end_date_price DECIMAL(10, 2) COMMENT 'Product price on the end date on the market.',
    price_currency STRING COMMENT 'Currency for product price.',
    market_name STRING COMMENT 'Market Name.',
    city STRING COMMENT 'City name where the Market is located.',
    department STRING COMMENT 'Department name where the city Market is located.',
    start_date DATE COMMENT 'Market assessment start date.',
    end_date DATE COMMENT 'Market assessment end date.',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
  ) USING DELTA
  CLUSTER BY (market_name, product_name, start_date)
  TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
  COMMENT 'Product List Silver Layer (data cleaned and validated)';
```
