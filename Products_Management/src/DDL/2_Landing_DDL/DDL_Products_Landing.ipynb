```python
%sql
DROP TABLE IF EXISTS products.landing.product_list_markets_landing;

create table products.landing.product_list_markets_landing (
    product_list_id BIGINT
      GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1)
      PRIMARY KEY
      COMMENT 'Auto incremental primary key for product market item',
    nombre_mercado STRING COMMENT 'market name',
    nombre_producto STRING COMMENT 'product name',
    unidad_medida STRING COMMENT 'unit measure',
    inicio_fecha_precios STRING COMMENT 'price start date',
    fin_fecha_precios STRING COMMENT 'price end date',
    precio_inicio STRING COMMENT 'initial product price',
    precio_fin STRING COMMENT 'final product price',
    error STRING COMMENT 'error reading documents',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
  ) USING DELTA
  TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
  COMMENT 'Table of Processed Product Price List Documents';
```
