# **FACT DIMENSION**
DDL for Products Fact Table in Gold Layer

```python
%sql
DROP TABLE IF EXISTS products.gold.fact_product_prices;

create table products.gold.fact_product_prices (
    product_hash STRING not null COMMENT 'Hash ID for Product Listing',
    product_id BigInt not null COMMENT 'Fk for Dim_Product',
    market_id BigInt not null COMMENT 'Fk for Dim_Market',  
    date_id BigInt not null COMMENT 'Fk for Dim_Time',
    start_date_price DECIMAL(10, 2) COMMENT 'Product price on the start date on the market',
    end_date_price DECIMAL(10, 2) COMMENT 'Product price on the end date on the market',
    price_currency STRING COMMENT 'Currency for product price',
    is_current Boolean default True COMMENT 'Is the product price valid',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
  ) USING DELTA
  CLUSTER BY (product_id,market_id)
  TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
  COMMENT 'Product Table Facts - Star Schema PK = row_hash (MD5 [product_id+id_date+market_id])';
```
