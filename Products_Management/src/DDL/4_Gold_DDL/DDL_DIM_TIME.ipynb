# **TIME DIMENSION**

DDL for Time Dimension Table in Gold Layer

```python
%sql
DROP TABLE IF EXISTS products.gold.dim_time;

create table products.gold.dim_time (
    date_id BIGINT COMMENT 'Measurement Date ID',
    start_market_date DATE NOT NULL COMMENT 'Start of Market Measurement Date',
    end_market_date DATE NOT NULL COMMENT 'End of Market Measurement Date',
    market_year INT COMMENT 'Market Measurement Year',
    market_month INT COMMENT 'Market Measurement Month',
    quarter INT COMMENT 'Market Measurement Quarter',
    semester INT COMMENT 'Market Measurement Semester',
    week_number INT COMMENT 'Week of the year of Market Measurement',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
  ) USING DELTA
  TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
  COMMENT 'Time Dimension - Star Schema';
```
