# **Original Product Document Price List**
DDL for Control Table for procesed documents in Bronze Layer

```python
%sql
DROP TABLE IF EXISTS products.bronze.docs_name;

create table products.bronze.docs_name (
    doc_id BIGINT
      GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1)
      PRIMARY KEY
      COMMENT 'Auto incremental primary key for read docs',
    docs_name STRING COMMENT 'docs name path',
    _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
  ) USING DELTA
  TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
  COMMENT 'Table of Processed Product Price List Documents';
```
